```python
import pandas as pd
from pydantic import BaseModel

class PredictionInput(BaseModel):
    total_amount: float
    hospital_bills: float
    claim_limits: float
    premium_amount: float
    treatment_expenses: float
    covered: str
    claim_documents_submitted: str
    fraud_history_approval_rejection_status: str
    benefits: str
    billing_frequency: str
    policy_type: str
    provider_id: str
    patient_id: str
    doctor: str
    hospital: str
    contact_details: str
    diagnosis_report: str
    discharge_summary: str
    prescriptions_and_bills: str
    insurance_company_name: str
    policy_number: str
    email: str
    address: str
    phone_number: str
    policy_name: str
    procedure_codes_cpt_code: str
    network_partners: str
    bank_account: str
    policy_holder_name: str
    start_date: str
    end_date: str
    renewal_date: str
    hospitalized_date: str

# Read CSV file (ensure you replace 'your_file.csv' with the actual file name)
df = pd.read_csv(r"C:\Bits Plani\CAPSTONE\Project_Work\code\main\insurance-fraud-anomaly-analysis\notebooks\conver_data.csv")
df.columns = df.columns.str.replace(' ', '_').str.replace('/', '_').str.lower()
print(df.head().T)
print(df.shape)

for _, row in df.iterrows():
    print(row.to_dict())  # Make sure all fields are properly structured

```

```python
# Convert CSV data into PredictionInput format
prediction_inputs = [PredictionInput(**row.to_dict()) for _, row in df.iterrows()]

# Print each JSON object
for prediction_input in prediction_inputs:
    print(prediction_input.model_dump_json())
```
